#### Align the NER labels to the tokenized inputs having special tokens and subwords.

```python
label_names = ds["train"].features["ner_tags"].feature.names
label_names # BOI NER tagging
```

Our goal is to align the labels for special tokens and also for sub-words.

Lets say for example: lamb is tokenized to `la` &  `##mb`. If the NER tag for word lamb is say `B-MISC`, then we have to make sure that for subwords `la` to be its original label and `##mb` to have NER tag as `I-MISC`. Remember to IOB and BOI Tagging.

We can achieve this with the help of `word_ids` and `labels`. Observe the order of label_names, every I-<tag> is just +1 of B-<tag>.

Also, we have to make sure that our special tokens are to be mapped to `-100` which is the default ignore index in hugging face.

```python
def align_labels_to_tokens(gt_labels, word_ids):
    new_labels = []
    current_word_idx = None

    for word_idx in word_ids:  # [None, 0, 1, 2, 3, 4, 5, 6, 7, 7, 8, None]
        if current_word_idx != word_idx:
            current_word_idx = word_idx
            label = -100 if word_idx is None else gt_labels[word_idx]
            new_labels.append(label)
        elif word_idx is None:
            new_labels.append(-100)  # -100 is a default index that HF will ignore
        else:  # current_word == word for subword i.e if the current token is from the same word as the previous token
            label = gt_labels[word_idx]
            if label % 2 == 1: # Check if the label is the odd i.e a beginining B-xxx token
                label += 1 # I is +1 of B. see the label_names. If the label is B-XXX, we change it to I-XXX for subwords
            new_labels.append(label)
    return new_labels
```

Note: A common convention (often used by Hugging Face) is to assign:
- Even numbers for O and I- tags.
- Odd numbers for B- tags.

For eg, see the labels:
- O = 0
- B-PER = 1
- I-PER = 2
- B-ORG = 3
- I-ORG = 4
- B-LOC = 5
- I-LOC = 6
- B-MISC = 7
- I-MISC = 8

label % 2 == 1 is used to detect if the original word's label was a "Beginning" (B-XXX) tag (odd). If it was, and the current token is a subsequent subword of that same original word, the code transforms the B-XXX tag into an I-XXX tag by simply adding 1 to its numerical ID.

```python
# Test
gt_labels = ds["train"][0]["ner_tags"]
word_ids = inputs.word_ids()  # tokenized inputs
print(gt_labels, word_ids)

print(align_labels_to_tokens(gt_labels, word_ids))
```

```python
def tokenize_and_align_labels(examples):
    tokenized_inputs = tokenizer(examples["tokens"], is_split_into_words=True, truncation=True)
    batch_gt_labels = examples["ner_tags"]
    new_batch_labels = []

    for idx, gt_labels in enumerate(batch_gt_labels):
        word_ids = tokenized_inputs.word_ids(idx)
        new_batch_labels.append(align_labels_to_tokens(gt_labels, word_ids))

    tokenized_inputs["labels"] = new_batch_labels
    return tokenized_inputs

tokenized_ds = ds.map(tokenize_and_align_labels, batched=True, remove_columns=ds["train"].column_names)
tokenized_ds
```

### Training

```python
from huggingface_hub import notebook_login, HfApi, create_repo, upload_folder, get_full_repo_name # For push the training model to hf like wandb to facilitate resume in case of crash
import os
from google.colab import userdata

from torch.optim import AdamW
from transformers import get_scheduler
from accelerate import Accelerator
from tqdm.auto import tqdm
import evaluate
import numpy as np

os.environ["TOKENIZERS_PARALELISM"] = "true"
# notebook_login() # logs to hf. can also use !huggingface-cli login in terminal/code cell



# Initializes a local Git repository that is to be linked and pushed to a remote repository on the Hugging Face Hub
output_dir = "bert-finetuned-ner-accelerate"  # name of the local directory where your model's files (weights, tokenizer, configuration, etc.) will be saved. It's common practice to use a local directory name that matches the model name for consistency.
os.makedirs(output_dir, exist_ok=True)
hf_token = userdata.get("HF_TOKEN")
api = HfApi(token=hf_token)

model_name = "bert-finetuned-ner-accelerate"
repo_id = get_full_repo_name(model_name, token=api.token) # hfusername/model_name
print(repo_id)
try:
    create_repo(repo_id=repo_id, repo_type="model", private=False, exist_ok=True, token=api.token)
except Exception as e:
    print(f"Could not create/check repo {repo_id}: {e}")

optimizer = AdamW(model.parameters(), lr=2e-5)
accelerator = Accelerator()

num_epochs = 3
num_batches = len(train_dataloader)
num_training_steps = num_epochs * num_batches
progress_bar = tqdm(range(num_training_steps))

lr_scheduler = get_scheduler("linear", optimizer=optimizer, num_warmup_steps=0, num_training_steps=num_training_steps)

train_dataloader, eval_dataloader, model, optimizer = accelerator.prepare(train_dataloader, eval_dataloader, model, optimizer)

metric = evaluate.load("seqeval")

# Post processing - Convert predictions and labels into strings and ignore the special tokens labels, like what our metric object expects
def post_process(predictions, gt_labels, label_names):
    predictions = predictions.detach().cpu().clone().numpy()
    gt_labels = gt_labels.detach().cpu().clone().numpy()
    preds = [[label_names[pred] for (label, pred) in zip(labels, preds) if label != -100] for labels, preds in zip(gt_labels, predictions)]
    gt_labels = [[label_names[label] for label in labels if label != -100] for labels in gt_labels]
    return gt_labels, preds

# Training
for epoch in range(num_epochs):
    model.train()
    for batch in train_dataloader:
        optimizer.zero_grad()
        outputs = model(**batch)
        loss = outputs.loss
        accelerator.backward(loss)
        optimizer.step()
        lr_scheduler.step()
        progress_bar.update(1)

    model.eval() # all batchnorm, dropout will be turned off....
    for batch in eval_dataloader:
        with torch.no_grad():
            outputs = model(**batch)
        logits = outputs.logits
        # print(logits.shape)  # (bs, max_seq_len_in_batch, 9)
        predictions = torch.argmax(logits, dim=-1)  # # (bs, max_seq_len_in_batch)
        gt_labels = batch["labels"] # (bs, max_seq_len)

        # Mandatory to pad predictions and labels to same length...if not...crash...
        predictions = accelerator.pad_across_processes(predictions, dim=1, pad_index=-100)
        gt_labels = accelerator.pad_across_processes(gt_labels, dim=1, pad_index=-100)

        # Gather from all the processes
        predictions_gathered = accelerator.gather(predictions)
        gt_labels_gathered = accelerator.gather(gt_labels)

        # Post process
        gt_labels, preds = post_process(predictions_gathered, gt_labels_gathered, label_names)
        metric.add_batch(predictions=preds, references=gt_labels)
    results = metric.compute()
    print(
        f"Epoch: {epoch} ",
        {key: results[f"overall_{key}"] for key in ["precision", "recall", "f1", "accuracy"]}
    )

    # save and upload the model
    accelerator.wait_for_everyone()
    unwrapped_model = accelerator.unwrap_model(model)  # start from scratch
    unwrapped_model.save_pretrained(output_dir, save_function=accelerator.save)

    if accelerator.is_main_process:
        tokenizer.save_pretrained(output_dir)
        try:
            upload_folder(repo_id=repo_id, folder_path=output_dir, repo_type="model", commit_message=f"Training in progress epoch {epoch}", token=api.token)
        except Exception as e:
            print(f"Failed to upload files: {e}")
```
